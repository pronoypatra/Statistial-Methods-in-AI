## 3.1 Hierarchical Clustering: Linkages and Features

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.datasets import make_blobs
```

```python
class hierarchical_clustering():
    def __init__(self):
        return

    def linkage_matrix(self, dataset, linkage_type, dist_metric):
        Z = linkage(dataset, method = linkage_type, metric = dist_metric)
        return Z

    def dendogram(self, Z):
        dendrogram(Z)
        return
```

```python
def hc(X):
    # Define linkage methods and distance metrics to explore
    linkage_methods = ['single', 'complete', 'average']
    distance_metrics = ['euclidean', 'cityblock', 'cosine']
    
    # Create subplots for each combination of linkage and distance metric
    fig, axes = plt.subplots(len(linkage_methods), len(distance_metrics), figsize=(15, 10))

    hier_clustering = hierarchical_clustering()
    
    for i in range (0, len(linkage_methods)):
        for j in range (0, len(distance_metrics)):
            # Linkage Matrices 
            lm = hier_clustering.linkage_matrix(X, linkage_methods[i], distance_metrics[j])
            print("Linkage Matrix:")
            print(lm)
            
            # Plot dendrogram
            ax = axes[i, j]
            dendrogram(lm, ax=ax)
            ax.set_title("Linkage: {}, Metric: {}".format(linkage_methods[i], distance_metrics[j]))
            ax.set_xlabel('Sample Index')
            ax.set_ylabel('Distance')
    
    plt.tight_layout()
    plt.show()
    
```

```python

df = pd.read_csv("./SMAI-Dataset-hc-dataset/new_customers.csv")

print("Unique Customer ID", df['CustomerID'].unique().shape)
print("Unique Age", df['Age'].unique().shape)
print("Unique Annual Income", df['Annual Income (k$)'].unique().shape)
print("Unique Spending Score", df['Spending Score (1-100)'].unique().shape)

df.drop(columns=['CustomerID'], inplace=True)
df.replace(['Male', 'Female'], [0,1], inplace=True)

data = df.to_numpy()

hc(data)
```
